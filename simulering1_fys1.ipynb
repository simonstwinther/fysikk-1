## Iterasjon i tid:
Når vi gjennomfører simuleringer i fysikk, har vi ofte noe informasjon om hvordan fysiske størrelser vil endre seg over små tidssteg. Vi kan bruke dette til å beregne verdien av størrelsene tidssteg for tidssteg, ved å itere gjennom tid. 

Vi skal begynne med et eksempel med konstant fart. Vi vet at
$v=\frac{\Delta s}{\Delta t}$, som betyr at endringen i posisjon er gitt av $\Delta s = v\cdot \Delta t$.
Vi kan bruke dette til å beregne endringen i posisjon for små tidssteg $\Delta t$, gjøre beregningen for mange små tidssteg, og slik finne endringen i posisjon over lengre tid.


```python
# Programmet beregner tilbakelagt strekning for 
# en gjenstand som beveger seg med konstant fart.

# Informasjon om bevegelsen
v = 5.0 #farten til gjenstanden, m/s
s = 0.0 #startposisjon, m
t = 0.0 #starttid, s

# Velger størrelse på tidssteg
# og hvor lenge vi ønsker å simulere
dt = 0.1 #lengde på tiddsteg, s
t_slutt = 3.0 #sluttid, s

# Her itererer vi gjennom tid, 
# beregner ny posisjon for hvert tidssteg

while t < t_slutt:
    s = s + v*dt #beregner ny posisjon (gammel pos + delta s)
    t = t + dt #tiden økes med dt

# Skriver ut resultatet
print("t =", t, " s =", s, " v =", v)
```

*Litt videre forklaring av koden:*

I while-løkken *itererer* vi oss gjennom tiden, altså øker $t$ med små tiddsteg $dt=0.1$ s for hver runde i løkken. Samtidig endrer vi posisjonen, som gitt av formelen 

$s_{ny}=s_{gammel}+\Delta s = s_{gammel}+v\cdot\Delta t $.

Vi bruker altså posisjonen vi står i ($s_{gammel}$) og ledder til endringen i posisjon, for å finne den nye posisjonen.

I den siste linja skriver vi ut hver av verdiene, etter at hele simuleringen er gjennomført. Den totale endringen i posisjon er altså 15 meter.

## Konstant akselerasjon:
Nå skal vi legge til en konstant akselerasjon $a=2,0 m/s^2$ i simuleringen vår. Dette betyr at også farten vil endre seg for hvert tidssteg. Vi kan bruke fartslikningen til å finne endringen i fart for hver iterasjon, som gitt av 
$\Delta v=a\cdot\Delta t$. I kode blir dette altså 

$v = v + a*dt$


**OPPGAVE:**
 - Legg til en linje i starten av programmet, som setter en konstant akselerasjon.
 - Legg til en linje i først i while-løkken som oppdaterer farten som gitt av den konstante akselerasjonen.

```python
# Programmet beregner tilbakelagt strekning for 
# en gjenstand som beveger seg med konstant akselerasjon.

# Informasjon om bevegelsen
v = 5.0 #startfarten til gjenstanden, m/s
s = 0.0 #startposisjon, m
t = 0.0 #starttid, s
# LEGG TIL KONSTANT AKSELERASJON

# Velger størrelse på tidssteg
# og hvor lenge vi ønsker å simulere
dt = 0.1 #lengde på tiddsteg, s
t_slutt = 3.0 #sluttid, s

# Her itererer vi gjennom tid, 
# beregner ny posisjon for hvert tidssteg

while t < t_slutt:
    # BEREGN NY FART
    s = s + v*dt #beregner ny posisjon (gammel pos + delta s)
    t = t + dt #tiden økes med dt

# Skriver ut resultatet
print("t =", t, " s =", s, " v =", v)
```

## Tegne grafer:

Koden under er en oppdatert versjon av det vi har jobbet med hittil, som tegner en graf som viser bevegelsen. Dette gjøres ved å lagre verdiene vi beregner for hvert tidssteg i lister, og så tegne grafer der vi bruker listen med tid $t$ som x-verdi, og listen med strekning $s$ som y-verdi.

```python
from pylab import *

# Programmet beregner tilbakelagt strekning for 
# en gjenstand som beveger seg med konstant akselerasjon.

# Informasjon om bevegelsen
a = 2.0 # akselerasjon, m/s^2
v = 5.0 #startfarten til gjenstanden, m/s
s = 0.0 #startposisjon, m
t = 0.0 #starttid, s


# Velger størrelse på tidssteg
# og hvor lenge vi ønsker å simulere
dt = 0.1 #lengde på tiddsteg, s
t_slutt = 3.0 #sluttid, s
t_verdier = [t] #lager en liste med t, legger inn første verdi
s_verdier = [s] #liste med s, legger inn første s

# Her itererer vi gjennom tid, 
# beregner ny posisjon for hvert tidssteg
while t < t_slutt:
    v = v + a*dt #beregner ny fart
    s = s + v*dt #beregner ny posisjon (gammel pos + delta s)
    t = t + dt #tiden økes med dt
    t_verdier.append(t) #legger inn ny verdi for tid i listen
    s_verdier.append(s) #legger inn ny verdi for s i listen

# Tegner graf
plot(t_verdier, s_verdier) #lager grafen
title("Strekning som funksjon av tid") #tittel
xlabel("$t$ / s") #navn på x-akse
ylabel("$s$ / m") #navn på y-akse
grid() #rutenett
show() #viser grafen
```

## Variabel akselerasjon
Endelig har vi kommet frem til punktet der vi kan løse oppgaver som bevegelseslikningene våre ikke kan løse: situajsoner med variabel akselerasjon.

Vi baserer oss på eksempel 22 i boka, for en 100 m-sprinter.
Vi har altså en situasjon der en sprinter har startakselerasjon $a=6,0 m/s^2$. Så tenker vi oss at akselerasjonen reduseres etter som farten øker. Vi setter opp en sammenheng mellom fart og akselerasjon, gitt som 


$a(v)=-0,50s^{-1}\cdot v+6,0m/s^2$. 

(Drøft kort, hva betyr enheten $s^{-1}$?).

Vi må beregne denne akselerasjonen for hvert tidssteg i bevegelsen. Hvordan kan vi gjøre dette? Vi kjenner farten $v$ ved hvert tidssteg, så kanskje vi ganske enkelt bare kan bruke funksjonen over til å beregne akselerasjonen i hver runde av løkken?

**OPPGAVE**: Legg inn funksjonen som beregner akselerasjon i den oppdaterte koden under.

```python
from pylab import *

# Programmet beregner tilbakelagt strekning for 
# en gjenstand som beveger seg med konstant akselerasjon.



# Informasjon om bevegelsen
v = 0.0 #startfart, m/s
s = 0.0 #startposisjon, m
t = 0.0 #starttid, s


# Velger størrelse på tidssteg
# og hvor lenge vi ønsker å simulere
dt = 0.01 #lengde på tiddsteg, s
s_slutt = 100.0 #sluttposisjon, m
s_verdier = [s] #liste med posisjon, legger inn første s
v_verdier = [v] #lager en liste med fart, legger inn første verdi


#Definerer funksjonen som beregner akselerasjonen
def a(v):
    return -0.5*v + 6.0 #LEGG INN FUNKSJONEN VI SKAL BEREGNE HER


# Her itererer vi gjennom tid, 
# beregner ny posisjon for hvert tidssteg
while s < s_slutt:
    v = v + a(v)*dt #OBS: MERK AT VI BRUKER FUNKSJONEN HER
    s = s + v*dt #beregner ny posisjon (gammel pos + delta s)
    t = t + dt #tiden økes med dt
    s_verdier.append(s) #legger inn ny verdi for posisjon i listen
    v_verdier.append(v) #legger inn ny verdi for fart i listen

# Tegner graf
plot(s_verdier, v_verdier) #lager grafen
title("Fart som funksjon av posisjon") #tittel
xlabel("$s$ / m") #navn på x-akse
ylabel("$v$ / (m/s)") #navn på y-akse
grid() #rutenett
show() #viser grafen
```
