```python
from pylab import *

def f(x):
    return e**(2*x) + x**2

def df_analytisk(x):
    return 2*e**(2*x) + 2*x

def df_numerisk(f, x, delta_x):
    return (f(x+delta_x)-f(x)) / delta_x

x = 2
delta_x = 0.1

#Her bruker jeg en funksjonalitet i print() som lar meg kombinere tekst og variabler.
#Der det står {variabel}, vil verdien av variabelen skrives ut.
#Der det står {variabel:.4f} vil det skrives ut et desimaltall med 4 tall etter komma.
print(f'Analytisk, f\'({x}) = {df_analytisk(x):.4f}')
print(f'Numerisk, f\'({x}) = {df_numerisk(f, x, delta_x):.4f}')
```

## Numerisk derivasjon:
Vi bruker definisjonen av den deriverte, 

$f'(x)=\lim_{\Delta x\rightarrow 0}\frac{f(x+\Delta x)-f(x)}{\Delta x}$

til å regne ut en tilnærmingsverdi av den deriverte i Python:

$f'(x)\approx\frac{f(x+\Delta x)-f(x)}{\Delta x}$


Dette kalles **numerisk derivasjon**, og uttrykket over kalles **Newton's kvotient**. 

Når vi deriverer med derivasjonsreglene, kalles det **analytisk derivasjon**.
